```python
import pandas as pd 
from dateutil.parser import parse 
# Sample data 
 
 
 
55 
data = { 
    'Event ID': ['E-001', 'E-002', 'E-003', 'E-004', 'E-005'], 
    'Event Date': ['01/12/2023', 'December 2, 2023', '2023-12-03', '12.04.2023', '5th Dec 2023'], 
    'Product Name': ['apple', 'Banana', 'ORANGE', 'BaNAna', 'Apple'], 
    'Weight' : ['70 kg', '100 lbs', '88 KG', '180 pound', '94 kilograms'] 
} 
df = pd.DataFrame(data) 
print("Original Dataset") 
print(df, '\n') 
```

```python
df
```

```python
def parse_date(date_str): 
    try: 
        return parse(date_str, dayfirst=False) 
    except ValueError: 
        return pd.NaT 
# Apply the custom function 
df['Event Date'] = df['Event Date'].apply(parse_date) 
# Standardize format to YYYY-MM-DD 
df['Event Date'] = df['Event Date'].dt.strftime('%Y-%m-%d') 
print("Result") 
print(df) 
```

```python
import numpy as np
import pandas as pd 
df=pd.DataFrame({
    "name":["abel","sara",None,"john"],
    "age":[20,np.nan,25,30],
    "salary":[4000,4500,5000,45000],
    "join_date":["2026-01-01","01/02/2026","march 10,2026","2026.04.15"]
})
```

```python
df
```

```python
df["name"]=df["name"].fillna("ayele")
```

```python
df["age"]=df["age"].fillna(df["age"].mean())
```

```python
df["join"]
```

```python
def parse_date(date_str): 
    try: 
        return parse(date_str, dayfirst=False) 
    except ValueError: 
        return pd.NaT 
df["join_date"]=df["join_date"].apply(parse_date)
df["join_date"]=df["join_date"].dt.strftime('%Y-%m-%d') 
```

```python
df
```

```python
plt.boxplot(df["salary"],vert=False)
plt.title("Outlire detection")
plt.xlabel("Salary($)")
plt.show()
```

```python
df["name"]=df["name"].str.capitalize()
```

```python
df
```

```python
data=df["salary"]
```

```python
q1=data.quantile(0.25)
q3=data.quantile(0.75)
iqr=q3-q1
```

```python
lb = q1-1.5*iqr
ub = q3+1.5*iqr
```

```python
data=df["salary"]
df["outlier"]=(data<lb)|(data>ub)
```
