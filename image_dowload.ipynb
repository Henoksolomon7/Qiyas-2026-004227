```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
html_doc=requests.get("https://www.bbc.com/news/articles/c78kmj2kyklo")
soup = BeautifulSoup(html_doc.content, 'html.parser')
articles=soup.find_all('article')
paragraph=soup.find_all('p')

os.makedirs("images", exist_ok=True)
```

```python
for i, img in enumerate(soup.find_all("img"), start=1):
    img_url = img.get("src")
for article in articles:
    time_tag = article.find("time")
    head_tag=article.find("h1")
    
    if head_tag:
        head=head_tag.get_text(strip=True)
        print(head)
    if time_tag:
        time=time_tag.get_text(strip=True)
        print(time)
    if articles:
     study = articles[0].find('div', class_="sc-8d737bf8-7 eGaKJA")
     print(study.text)
    #if article:
        #image=articles[0].find('img')
        #print(image.get('src'))
    for i, img in enumerate(article.find_all("img"), start=1):
     img_url = img.get("src")   
     if img_url:
        try:
            img_data = requests.get(img_url).content

            with open(f"pic/image_{i}.jpg", "wb") as f:
                f.write(img_data)

            print(f"Downloaded image {i}")
        except Exception as e:
            print(f"Failed: {e}")
    for p in paragraph:
     text = p.get_text(strip=True)

     if text:
        print(text)
       #print(f"{head:<12} {time:<12}{text:<120}")
    
        
    #print(article.get_text(strip=True))
```
