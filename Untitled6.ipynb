```python
np.random.seed(0)
sales_amounts=np.random.randint(20, size=(5,3))
sales_amounts
```

```python
sales_amounts
```

```python
#create weekly_sales DataFrame
import pandas as pd
weekly_sales=pd.DataFrame(sales_amounts,
                          index=["mondau","tusday","wendsday","thursday","friday"],
                          columns=["Almond butter","peanut butter","cashew butter"])
```

```python
weekly_sales
```

```python
price=np.array([10,8,12])
```

```python
price.shape
```

```python
butter_prices=pd.DataFrame(price.reshape(1,3),
                          index=["price"],
                          columns=["Almond butter","peanut butter","cashew butter"])
butter_prices
```

```python
total_sales=np.dot(sales_amounts,price)
```

```python
total_sales
```

```python
#
weekly_sales["total price($)"]=total_sales.T
weekly_sales
```
