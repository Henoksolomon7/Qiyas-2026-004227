```python
from dash import Dash, dcc, html, Input, Output, callback

external_stylesheets = ['https://codepen.io/chriddyp/pen/bWLwgP.css']

app = Dash(__name__, external_stylesheets=external_stylesheets)

all_options = {
    'America': ['New York City', 'San Francisco', 'Cincinnati'],
    'Canada': ['Montréal', 'Toronto', 'Ottawa']
}
app.layout = html.Div([
    dcc.RadioItems(
        list(all_options.keys()),
        'America',
        id='countries-radio',
    ),

    html.Hr(),

    dcc.RadioItems(id='cities-radio'),

    html.Hr(),

    html.Div(id='display-selected-values')
])


@callback(
    Output('cities-radio', 'options'),
    Input('countries-radio', 'value'))
    
def set_cities_options(selected_country):
    return [{'label': i, 'value': i} for i in all_options[selected_country]]


@callback(
    Output('cities-radio', 'value'),
    Input('cities-radio', 'options'))
def set_cities_value(available_options):
    return available_options[0]['value']


@callback(
    Output('display-selected-values', 'children'),
    Input('countries-radio', 'value'),
    Input('cities-radio', 'value'))
def set_display_children(selected_country, selected_city):
    return f'{selected_city} is a city in {selected_country}'


if __name__ == '__main__':
    app.run(debug=True,port=8054)
```

```python
from dash import Dash, html, dcc, Input, Output, dash_table
import plotly.express as px
import pandas as pd

# -----------------------------
# App
# -----------------------------
app = Dash(__name__)

# Make sure fana_data already exists
# fana_data must contain:
# language, category, Author, date, title

fana_data["date"] = pd.to_datetime(fana_data["date"], errors="coerce")

# -----------------------------
# Layout
# -----------------------------
app.layout = html.Div([

    html.H1("Fana News Analytics Dashboard"),

    # Filters
    html.Div([

        dcc.Dropdown(
            id="language",
            options=[{"label": i, "value": i} for i in fana_data["language"].dropna().unique()],
            value=fana_data["language"].dropna().unique()[0],
            clearable=False
        ),

        dcc.Dropdown(
            id="category",
            options=[{"label": i, "value": i} for i in fana_data["category"].dropna().unique()],
            multi=True,
            placeholder="Select category"
        ),

        dcc.Input(
            id="search",
            type="text",
            placeholder="Search title...",
            style={"width": "100%", "marginTop": "10px"}
        ),

    ], style={"width": "50%"}),

    html.Br(),

    # KPI cards
    html.Div([
        html.Div(id="kpi_articles"),
        html.Div(id="kpi_authors"),
        html.Div(id="kpi_categories"),
    ], style={"display": "flex", "gap": "20px"}),

    html.Br(),

    # Charts
    dcc.Graph(id="line_chart"),
    dcc.Graph(id="bar_chart"),
    dcc.Graph(id="pie_chart"),

    html.Br(),

    # Table
    dash_table.DataTable(
        id="table",
        page_size=10,
        style_table={"overflowX": "auto"}
    )

])


# -----------------------------
# Callback (ALL IN ONE)
# -----------------------------
@app.callback(
    Output("kpi_articles", "children"),
    Output("kpi_authors", "children"),
    Output("kpi_categories", "children"),
    Output("line_chart", "figure"),
    Output("bar_chart", "figure"),
    Output("pie_chart", "figure"),
    Output("table", "data"),

    Input("language", "value"),
    Input("category", "value"),
    Input("search", "value")
)
def update_dashboard(language, category, search):

    df = fana_data.copy()

    # ---------------- filtering ----------------
    df = df[df["language"] == language]

    if category:
        df = df[df["category"].isin(category)]

    if search:
        df = df[df["title"].str.contains(search, case=False, na=False)]

    # ---------------- KPIs ----------------
    kpi_articles = f"Articles: {len(df)}"
    kpi_authors = f"Authors: {df['Author'].nunique()}"
    kpi_categories = f"Categories: {df['category'].nunique()}"

    # ---------------- Line chart ----------------
    line_df = df.groupby("date").size().reset_index(name="count")

    line_fig = px.line(
        line_df,
        x="date",
        y="count",
        title="Articles Over Time"
    )

    # ---------------- Author chart ----------------
    author_df = df["Author"].value_counts().reset_index()
    author_df.columns = ["Author", "count"]

    bar_fig = px.bar(
        author_df,
        x="count",
        y="Author",
        orientation="h",
        title="Top Authors"
    )

    # ---------------- Pie chart ----------------
    pie_fig = px.pie(
        df,
        names="category",
        title="Category Distribution"
    )
    df["source_url"] = df["source_url"].apply(
    lambda x: f"[Read Article]({x})"
           )
    # ---------------- Table ----------------
    table_data = df[["date", "title", "Author", "category", "language","source_url"]].to_dict("records")

    return (
        kpi_articles,
        kpi_authors,
        kpi_categories,
        line_fig,
        bar_fig,
        pie_fig,
        table_data
    )


# -----------------------------
# Run app
# -----------------------------
app.run(debug=True, port=8051)
```
