```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

datas=pd.read_csv(r"C:\Users\SOLOMON\Downloads\spam_Emails_data.csv\spam_Emails_data.csv")
datas['text']=datas['text'].fillna("you are a good person to trust")

# Convert text to numbers
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(datas['text'])
Y= datas['label']
# Train model
model = MultinomialNB()
model.fit(X,Y)

# Predict new email
message = input("Type an email: ")

message_vector = vectorizer.transform([message])

prediction = model.predict(message_vector)

print("\nPrediction:", prediction[0])
```
