```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd
import re
# from datetime import datetime 

BASE_URL="https://ethiopianbusinessreview.net/"

url = "https://ethiopianbusinessreview.net/category/featured-articles/"
headers = {
    "User-Agent": "Mozilla/5.0"
}
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
ul=soup.find(id="menu-primary-menu")
links=set()
for a in ul.find_all("a", href=True):
    href = a["href"]

    if "ethiopianbusinessreview.net" in href:
        links.add(urljoin(BASE_URL, href))
        #print(href)
print(f"Found {len(links)} articles")  
data = []
date = "No date"
for link in links:
    try:
        article_page= requests.get(link, headers=headers)
        article_soup = BeautifulSoup(article_page.text, "html.parser")

        title = article_soup.find("h2"or "h1")
        title = title.get_text(strip=True) if title else "No title"
       # time = article_soup.find("time")
       # time = time.get_text(strip=True)
        date_tag = article_soup.find(class_="btIconWidgetTitle")
#class_="btIconWidgetTitle"
        date = date_tag.get_text(strip=True) if date_tag else "No date"

        
        #print(date)

        
       # writer=article_soup.find("time")
        paragraphs = article_soup.find_all("p")
        
        content = " ".join(
            p.get_text(strip=True)
            for p in paragraphs
        )

        data.append({
            "title": title,
            "time":date,
            "url": link,
            "content": content
        })
        print("success!")
        print(title,":\n","Time:",date,"\n","link:",link,":\n", content)


    except Exception as e:
        print("Error:", e)

df = pd.DataFrame(data)
df.to_csv("fbs_technology_news.csv", index=False)
df.to_csv("fbs_technology_news.txt", index=False)

print("Saved!")
    
```
