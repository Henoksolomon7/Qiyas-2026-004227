```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 2. LOAD DATA
# ============================================================

# Change this to your actual file
df = pd.read_csv(r"C:\Users\SOLOMON\Downloads\Telegram Desktop\weatherdata.csv")

print("Dataset shape:", df.shape)
print(df.head())
print(df.info())



df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Sort chronologically
df = df.sort_values("Date").reset_index(drop=True)

# Remove rows where Date is invalid
df = df.dropna(subset=["Date"])


# ============================================================
# 4. CREATE TIME FEATURES
# ============================================================

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfYear"] = df["Date"].dt.dayofyear

# Cyclic seasonal features
df["Month_sin"] = np.sin(2 * np.pi * df["Month"] / 12)
df["Month_cos"] = np.cos(2 * np.pi * df["Month"] / 12)

df["DayOfYear_sin"] = np.sin(
    2 * np.pi * df["DayOfYear"] / 365.25
)

df["DayOfYear_cos"] = np.cos(
    2 * np.pi * df["DayOfYear"] / 365.25
)


# ============================================================
# 5. REMOVE UNNECESSARY COLUMNS
# ============================================================

# Change this to your actual target
target = "Precipitation"

# Date itself should not be passed directly into sklearn models
df = df.drop(columns=["Date"])


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

# Replace infinite values
df = df.replace([np.inf, -np.inf], np.nan)

# For a large weather dataset, median imputation is safer
# than simply deleting huge numbers of rows.
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)


# ============================================================
# 7. DEFINE X AND y
# ============================================================

X = df.drop(columns=[target])
y = df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)


# ============================================================
# 8. CHRONOLOGICAL TRAIN/TEST SPLIT
# ============================================================

# IMPORTANT:
# Do NOT randomly split time-series/weather data.

split_index = int(len(df) * 0.80)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


#============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Ridge": Ridge(alpha=1.0),

    "Lasso": Lasso(alpha=0.001, max_iter=10000),

    "KNN": KNeighborsRegressor(
        n_neighbors=5,
        n_jobs=-1
    ),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    )
}



def evaluate_models(scaler, scaler_name):

    results = []
    predictions = {}

    for name, model in models.items():

        print(f"\nTraining {name} using {scaler_name}...")

        try:

            # Pipeline prevents data leakage.
            # Scaler is fitted ONLY on training data.
            pipeline = Pipeline([
                ("scaler", scaler),
                ("model", model)
            ])

            # Train
            pipeline.fit(X_train, y_train)

            # Predict
            y_pred = pipeline.predict(X_test)

            # Metrics
            mae = mean_absolute_error(y_test, y_pred)

            rmse = np.sqrt(
                mean_squared_error(y_test, y_pred)
            )

            r2 = r2_score(y_test, y_pred)

            results.append({
                "Scaler": scaler_name,
                "Model": name,
                "MAE": mae,
                "RMSE": rmse,
                "R2 Score": r2
            })

            predictions[f"{name} - {scaler_name}"] = y_pred

            print(
                f"MAE: {mae:.4f} | "
                f"RMSE: {rmse:.4f} | "
                f"R²: {r2:.4f}"
            )

        except Exception as e:

            print(
                f"{name} failed with {scaler_name}: {e}"
            )

    return results, predictions



standard_results, standard_predictions = evaluate_models(
    StandardScaler(),
    "StandardScaler"
)



minmax_results, minmax_predictions = evaluate_models(
    MinMaxScaler(),
    "MinMaxScaler"
)



results = standard_results + minmax_results

results_df = pd.DataFrame(results)



results_df = results_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)


print("\n==============================")
print("MODEL COMPARISON")
print("==============================")

print(
    results_df.to_string(index=False)
)


# ============================================================
# 15. BEST MODEL
# ============================================================

best_model = results_df.iloc[0]

print("\n==============================")
print("BEST MODEL")
print("==============================")

print("Scaler :", best_model["Scaler"])
print("Model  :", best_model["Model"])
print("MAE    :", best_model["MAE"])
print("RMSE   :", best_model["RMSE"])
print("R²     :", best_model["R2 Score"])
```

```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 2. LOAD DATA
# ============================================================
df=pd.read_csv(r"C:\Users\SOLOMON\Downloads\Telegram Desktop\weatherdata.csv")


print("Original shape:", df.shape)
print(df.head())


# ============================================================
# 3. DATE PROCESSING
# ============================================================

df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)

# Remove invalid dates
df = df.dropna(subset=["Date"])

# Sort chronologically
df = df.sort_values("Date").reset_index(drop=True)


# ============================================================
# 4. CREATE TIME FEATURES
# ============================================================

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfYear"] = df["Date"].dt.dayofyear

# Seasonal/cyclic features
df["Month_sin"] = np.sin(
    2 * np.pi * df["Month"] / 12
)

df["Month_cos"] = np.cos(
    2 * np.pi * df["Month"] / 12
)

df["DayOfYear_sin"] = np.sin(
    2 * np.pi * df["DayOfYear"] / 365.25
)

df["DayOfYear_cos"] = np.cos(
    2 * np.pi * df["DayOfYear"] / 365.25
)


# ============================================================
# 5. TARGET
# ============================================================

target = "Precipitation"


# ============================================================
# 6. REMOVE DATE
# ============================================================

df = df.drop(columns=["Date"])


# ============================================================
# 7. HANDLE INFINITE VALUES
# ============================================================

df = df.replace(
    [np.inf, -np.inf],
    np.nan
)


# ============================================================
# 8. HANDLE MISSING VALUES
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)


# ============================================================
# 9. X AND y
# ============================================================

X = df.drop(columns=[target])
y = df[target]


print("\nX shape:", X.shape)
print("y shape:", y.shape)


# ============================================================
# 10. CHRONOLOGICAL TRAIN / TEST SPLIT
# ============================================================

split_index = int(len(df) * 0.80)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()


print("\nTraining rows:", len(X_train))
print("Testing rows :", len(X_test))


# ============================================================
# 11. MODELS
# ============================================================

models = {

    "Linear Regression": LinearRegression(),

    "Ridge": Ridge(
        alpha=1.0
    ),

    "Lasso": Lasso(
        alpha=0.001,
        max_iter=10000
    ),

    "KNN": KNeighborsRegressor(
        n_neighbors=5,
        n_jobs=-1
    ),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    )
}


# ============================================================
# 12. SCALERS
# ============================================================

scalers = {

    "No Scaler": None,

    "StandardScaler": StandardScaler(),

    "MinMaxScaler": MinMaxScaler()
}


# ============================================================
# 13. TRAIN AND COMPARE
# ============================================================

results = []

predictions = {}


for scaler_name, scaler in scalers.items():

    print("\n")
    print("=" * 70)
    print(f"        {scaler_name}")
    print("=" * 70)

    for model_name, model in models.items():

        print(
            f"Training {model_name} "
            f"with {scaler_name}..."
        )

        try:

            # ------------------------------------------------
            # CREATE PIPELINE
            # ------------------------------------------------

            if scaler is None:

                pipeline = Pipeline([
                    ("model", model)
                ])

            else:

                pipeline = Pipeline([
                    ("scaler", scaler),
                    ("model", model)
                ])


            # ------------------------------------------------
            # TRAIN
            # ------------------------------------------------

            pipeline.fit(
                X_train,
                y_train
            )


            # ------------------------------------------------
            # PREDICT
            # ------------------------------------------------

            y_pred = pipeline.predict(
                X_test
            )


            # ------------------------------------------------
            # METRICS
            # ------------------------------------------------

            mae = mean_absolute_error(
                y_test,
                y_pred
            )

            rmse = np.sqrt(
                mean_squared_error(
                    y_test,
                    y_pred
                )
            )

            r2 = r2_score(
                y_test,
                y_pred
            )


            # ------------------------------------------------
            # SAVE RESULTS
            # ------------------------------------------------

            results.append({

                "Scaler": scaler_name,

                "Model": model_name,

                "MAE": mae,

                "RMSE": rmse,

                "R2 Score": r2

            })


            # Save predictions
            predictions[
                f"{model_name} - {scaler_name}"
            ] = y_pred


            print(
                f"    MAE  = {mae:.4f}"
            )

            print(
                f"    RMSE = {rmse:.4f}"
            )

            print(
                f"    R²   = {r2:.4f}"
            )


        except Exception as e:

            print(
                f"ERROR in {model_name} "
                f"with {scaler_name}: {e}"
            )


# ============================================================
# 14. CREATE RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    results
)


# ============================================================
# 15. SORT BY RMSE
# ============================================================

results_df = results_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)


# ============================================================
# 16. DISPLAY ALL RESULTS
# ============================================================

print("\n")
print("=" * 80)
print("                    MODEL COMPARISON")
print("=" * 80)

print(
    results_df.to_string(index=False)
)


# ============================================================
# 17. BEST MODEL
# ============================================================

best_model = results_df.iloc[0]

print("\n")
print("=" * 80)
print("                         BEST MODEL")
print("=" * 80)

print(
    f"Scaler : {best_model['Scaler']}"
)

print(
    f"Model  : {best_model['Model']}"
)

print(
    f"MAE    : {best_model['MAE']:.4f}"
)

print(
    f"RMSE   : {best_model['RMSE']:.4f}"
)

print(
    f"R²     : {best_model['R2 Score']:.4f}"
)
```
