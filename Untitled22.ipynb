```python
import tensorflow as tf
from tensorflow.keras import layers, Model

inputs = layers.Input(shape=(256, 256, 3))

# Encoder
x = layers.Conv2D(64, 3, padding="same", activation="relu")(inputs)
x = layers.Conv2D(64, 3, padding="same", activation="relu")(x)

x = layers.MaxPooling2D()(x)

x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)
x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)

x = layers.MaxPooling2D()(x)

# Bottleneck
x = layers.Conv2D(256, 3, padding="same", activation="relu")(x)
x = layers.Conv2D(256, 3, padding="same", activation="relu")(x)

# Decoder
x = layers.UpSampling2D()(x)

x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)
x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)

x = layers.UpSampling2D()(x)

x = layers.Conv2D(64, 3, padding="same", activation="relu")(x)
x = layers.Conv2D(64, 3, padding="same", activation="relu")(x)

# Output
outputs = layers.Conv2D(3, 3, padding="same", activation="sigmoid")(x)

model = Model(inputs, outputs)

model.compile(
    optimizer="adam",
    loss="mse", metrics=["accuracy"]
)

model.summary()
```

```python
import glob
import cv2
import numpy as np

files = glob.glob("water_marked/*.jpg")

print(files)
print("Number of images:", len(files))
```

```python
file= glob.glob("clean/*.jpg")
```

```python
X = []

for file in files:
    image = cv2.imread(file)

    image = cv2.resize(image, (256, 256))

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    image = image / 255.0

    X.append(image)

X = np.array(X, dtype=np.float32)

print(X.shape)
```

```python
Y=[]
for fil in file:
    image = cv2.imread(fil)

    image = cv2.resize(image, (256, 256))

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    image = image / 255.0

    Y.append(image)

X = np.array(X, dtype=np.float32)

print(X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.1,
    random_state=42
)
```

```python
import numpy as np

X_train = np.array(X_train, dtype=np.float32)
Y_train = np.array(Y_train, dtype=np.float32)

X_test = np.array(X_test, dtype=np.float32)
Y_test = np.array(Y_test, dtype=np.float32)
```

```python
print("X_train:", X_train.shape)
print("Y_train:", Y_train.shape)

print("X_test:", X_test.shape)
print("Y_test:", Y_test.shape)
```

```python
history=model.fit(
    X_train,
    Y_train,
    validation_data=(X_test, Y_test),
    epochs=10,
    batch_size=16
)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'], label = 'train_accuracy')
plt.plot(history.history['val_accuracy'], label = 'val_accuracy')
plt.legend()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

inputs = layers.Input(shape=(256, 256, 3))

# --- Encoder ---
# Block 1
x1 = layers.Conv2D(64, 3, padding="same", use_bias=False)(inputs)
x1 = layers.BatchNormalization()(x1)
x1 = layers.Activation("relu")(x1)
x1 = layers.Conv2D(64, 3, padding="same", use_bias=False)(x1)
x1 = layers.BatchNormalization()(x1)
x1 = layers.Activation("relu")(x1)

p1 = layers.MaxPooling2D()(x1)

# Block 2
x2 = layers.Conv2D(128, 3, padding="same", use_bias=False)(p1)
x2 = layers.BatchNormalization()(x2)
x2 = layers.Activation("relu")(x2)
x2 = layers.Conv2D(128, 3, padding="same", use_bias=False)(x2)
x2 = layers.BatchNormalization()(x2)
x2 = layers.Activation("relu")(x2)

p2 = layers.MaxPooling2D()(x2)

# --- Bottleneck ---
b = layers.Conv2D(256, 3, padding="same", use_bias=False)(p2)
b = layers.BatchNormalization()(b)
b = layers.Activation("relu")(b)
b = layers.Conv2D(256, 3, padding="same", use_bias=False)(b)
b = layers.BatchNormalization()(b)
b = layers.Activation("relu")(b)

# --- Decoder (With Skip Connections) ---
# Block 2 Up
u2 = layers.Conv2DTranspose(128, 2, strides=2, padding="same")(b)
u2 = layers.Concatenate()([u2, x2])  # Skip connection from Block 2
u2 = layers.Conv2D(128, 3, padding="same", use_bias=False)(u2)
u2 = layers.BatchNormalization()(u2)
u2 = layers.Activation("relu")(u2)

# Block 1 Up
u1 = layers.Conv2DTranspose(64, 2, strides=2, padding="same")(u2)
u1 = layers.Concatenate()([u1, x1])   # Skip connection from Block 1
u1 = layers.Conv2D(64, 3, padding="same", use_bias=False)(u1)
u1 = layers.BatchNormalization()(u1)
u1 = layers.Activation("relu")(u1)

# --- Output ---
outputs = layers.Conv2D(3, 3, padding="same", activation="sigmoid")(u1)

Model = Model(inputs, outputs)

# Combination loss for sharper images (MSE + Structural Similarity)
def combined_loss(y_true, y_pred):
    mse = tf.reduce_mean(tf.square(y_true - y_pred))
    ssim = 1.0 - tf.reduce_mean(tf.image.ssim(y_true, y_pred, max_val=1.0))
    return mse + 0.5 * ssim

Model.compile(optimizer="adam", loss=combined_loss)
Model.summary()
```

```python
history=Model.fit(
    X_train,
    Y_train,
    validation_data=(X_test, Y_test),
    epochs=10,
    batch_size=16
)
```
