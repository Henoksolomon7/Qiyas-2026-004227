```python
data = { 
    'Student_ID': ['S001', 'S002', 'S003', 'S004', 'S005'], 
    'Mathematics' : [85, 75, 95, 65, 90], 
   'Science' : [78, 88, 82, 85, 79], 
    'English' : [90, 85, 80, 70, 95], 
    'Information Technology' : [89, 65, 88, 92, 98] 
} 
df_result = pd.DataFrame(data) 
df_result
```

```python
import numpy as np
df_result['Average'] = np.mean(df_result[['Mathematics', 'Science','English', 'Information Technology']], axis=1) 
df_result
```

```python
df_result['Weighted Average'] = (df_result['Mathematics'] * 0.4) + (df_result['Science'] * 0.2) + (df_result['English'] * 0.2) + (df_result['Information Technology'] * 0.2) 
df_result 
```
