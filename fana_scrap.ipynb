```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd

url = "https://www.fanamc.com/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

navigation = soup.find(id="top-navigation")

links = set()

for a in navigation.find_all("a", href=True):
    href = a["href"]

    if "fanamc.com" in href:
        links.add(urljoin(url, href))

print(f"Found {len(links)} links")

data = []

for link in links:
    try:
        article_page = requests.get(link, headers=headers)
        article_soup = BeautifulSoup(article_page.text, "html.parser")
        #language=article_soup(link)
        #language=language.get_text(strip=True) if title else "No language"
        title = article_soup.find(["h1", "h2"])
        title = title.get_text(strip=True) if title else "No title"

        date_tag = article_soup.find(class_="post-published updated")
        date = date_tag.get_text(strip=True) if date_tag else "No date"

        paragraphs = article_soup.find_all("p")

        content = " ".join(
            p.get_text(strip=True)
            for p in paragraphs
        )

        data.append({
            #"language":language,
            "title": title,
            "date": date,
            "url": link,
            "content": content
        })

        print("Success:", title)

    except Exception as e:
        print("Error:", e)

df = pd.DataFrame(data)

df.to_csv("fanas_news.csv", index=False)
df.to_csv("fanas_news.txt", index=False)

print("Saved!")

```
