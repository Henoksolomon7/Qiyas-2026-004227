```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd

BASE_URL = "https://www.fanamc.com/english/"
CATEGORY_URL = "https://www.fanamc.com/english/business/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

# -----------------------------
# Get the category page
# -----------------------------
response = requests.get(CATEGORY_URL, headers=headers)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

# -----------------------------
# Collect article links
# -----------------------------
links = set()

for a in soup.find_all("a", href=True):
    href = urljoin(BASE_URL, a["href"])

    if (
        href.startswith(BASE_URL)
        and "/category/" not in href
        and "/tag/" not in href
        and "/author/" not in href
        and href.rstrip("/") != BASE_URL.rstrip("/")
    ):
        links.add(href)

print(f"Found {len(links)} possible links")

# -----------------------------
# Scrape each article
# -----------------------------
data = []

for link in sorted(links):

    try:
        page = requests.get(link, headers=headers)
        page.raise_for_status()

        article = BeautifulSoup(page.text, "html.parser")

        # -------------------------
        # Title
        # -------------------------
        title_tag = article.find(class_="post-title")
        title = title_tag.get_text(strip=True) if title_tag else "No title"

        # -------------------------
        # Date
        # -------------------------
        date_tag = article.find(class_="post-published updated")

        if not date_tag:
            date_tag = article.find("time")

        date = date_tag.get_text(strip=True) if date_tag else "No date"

        # -------------------------
        # Author
        # -------------------------
        author_tag = article.find(class_="post-author author")

        if not author_tag:
            author_tag = article.find(rel="author")

        author = author_tag.get_text(strip=True) if author_tag else "No author"

        # -------------------------
        # Article Content
        # -------------------------
        content = ""

        #content_div = article.find(class_="entry-content clearfix single-post-content")
        #class_="entry-content clearfix single-post-content
        #content_div = article_soup.find("div", class_="entry-content")
        for div in article.find("div"):
             content_div = div.get("class")
             if content_div:
                print(content_div)
             if content_div:
               paragraphs = content_div.find_all("p")

               if paragraphs:
                content = " ".join(
                    p.get_text(" ", strip=True)
                    for p in paragraphs
                )
               else:
                content = content_div.get_text(" ", strip=True)

        if content == "":
            content = "No content"

        # Skip pages that aren't articles
        if title == "No title" and content == "No content":
            continue

        data.append({
            "language": "English",
            "category": "Business",
            "title": title,
            "date": date,
            "author": author,
            "article_summary": content,
            "source_url": link
        })

        print("=" * 80)
        print(title)
        print(date)
        print(author)
        print(link)
        print(content[:300], "...\n")

    except Exception as e:
        print(f"Error scraping {link}")
        print(e)

# -----------------------------
# Save CSV
# -----------------------------
#df = pd.DataFrame(data)

#df.to_csv("fana_english_business.csv", index=False, encoding="utf-8-sig")

print(f"\nSaved {len(df)} articles.")
```
