```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
html_doc=requests.get("https://www.worldometers.info/coronavirus/")
soup= BeautifulSoup(html_doc.content,'html.parser')
```

```python
table_content=soup.find("table")
table_head= table_content.find("thead")
table_body=table_content.find("tbody")
table_row=table_body.find_all("tr")
#for a in table_head:

headers=['Country','Total_case','New_case','Total_death','New_deth','Active_case']
print(f"{headers[0]:<35}{headers[1]:<15}{headers[2]:<15}{headers[3]:<15}{headers[4]:<15}{headers[5]:<15}")
with open("covi_file.csv","a" ,encoding="utf-8") as file:
      file.write(headers[0]+'#'+headers[1]+'#'+headers[2]+'#'+headers[3]+'#'+headers[4]+'#'+headers[5]+'#'+'\n')
      for data in table_row[8:]:
         cells=data.find_all('td')
         if len(cells)>1:
             country=cells[1].text.strip()
             total_case=cells[2].text.strip()
             new_case=cells[3].text.strip()
             total_death=cells[4].text.strip()
             new_death=cells[5].text.strip()
             active_case=cells[8].text.strip()
            # total_tests=cells[9].text.strip()
             Total_case=total_case.replace(","," ")
             Total_death=total_death.replace(","," ")
             Active_case=active_case.replace(","," ")
             
         print(f"{country:<35} {Total_case}{new_case: >15} {Total_death :>12}{new_death:>14}{Active_case:>15}")
         text = country+"#"+Total_case+"#"+new_case+"#"+Total_death +"#"+new_death+"#"+Active_case+'\n'
         file.write(text)
         
```
